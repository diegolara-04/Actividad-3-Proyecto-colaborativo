## 21. zip()

**Descripción:** La funcion zip() combina múltiples objetos iterables (como listas o tuplas) en un solo iterador de tuplas, emparejando los elementos según su posición (el primero con el primero, el segundo con el segundo, etc.).

**Sintaxis:** zip(iterable1, iterable2, ...)

**Ejemplo práctico:**

```python
nombres = ["Ana", "Luis", "Carlos"]
edades = [25, 30, 22]
combinados = list(zip(nombres, edades))
print(combinados) # Salida: [('Ana', 25), ('Luis', 30), ('Carlos', 22)]
```

## 22. map()

**Descripción:** Aplica una función específica a todos los elementos de un iterable y devuelve un nuevo iterador con los resultados.

**Sintaxis:** map(función, iterable)

**Ejemplo práctico:**

```python
numeros = [1, 2, 3, 4]
cuadrados = list(map(lambda x: x**2, numeros))
print(cuadrados) # Salida: [1, 4, 9, 16]
```

## 23. filter()

**Descripción:** Crea un nuevo iterador extrayendo solo aquellos elementos de un iterable original que cumplen con una condición o función específica (donde la función devuelve verdadero).

**Sintaxis:** filter(función, iterable)

**Ejemplo práctico:**

```python
edades = [15, 20, 17, 22, 30]
mayores_edad = list(filter(lambda x: x >= 18, edades))
print(mayores_edad) # Salida: [20, 22, 30]
```

## 24. any()


**Descripción:** Devuelve el valor True si al menos un elemento dentro de un iterable es evaluado como verdadero. Si el iterable está vacío, devuelve False.

**Sintaxis:** any(iterable)

**Ejemplo práctico:**

```python
conexiones_servidor = [False, False, True, False]
print(any(conexiones_servidor)) # Salida: True
```

## 25. all()


**Descripción:** Devuelve el valor True únicamente si todos los elementos dentro de un iterable son evaluados como verdaderos o si el iterable está vacío.

**Sintaxis:** all(iterable)

**Ejemplo práctico:**

```python
requisitos_cumplidos = [True, True, True]
print(all(requisitos_cumplidos)) # Salida: True
```

## 26. reversed()


**Descripción:** Toma un iterable (como una lista o un rango) y devuelve un nuevo iterador que contiene los mismos elementos pero en el orden inverso.

**Sintaxis:** reversed(secuencia)

**Ejemplo práctico:**

```python
letras = ['a', 'b', 'c', 'd']
al_reves = list(reversed(letras))
print(al_reves) # Salida: ['d', 'c', 'b', 'a']
```

## 27. open()


**Descripción:** Es la función principal para el manejo de archivos en Python. Abre un archivo del sistema y devuelve un objeto de archivo que permite leerlo, escribir en él o modificarlo según el modo de apertura indicado.

**Sintaxis:** open(archivo, modo)

**Ejemplo práctico:**

```python
with open("prueba_actividad.txt", "w") as archivo:
    archivo.write("Este es un archivo creado desde Python.")
```

## 28. help()


**Descripción:** Invoca el sistema de ayuda integrado de Python. Es útil para leer la documentación de objetos, módulos, funciones o clases directamente en la consola.

**Sintaxis:** help(objeto)

**Ejemplo práctico:**

```python
help(print) # Muestra la documentación oficial de la función print
```

## 29. insinstance()


**Descripción:** Comprueba si un objeto determinado es una instancia de una clase o tipo de dato específico (o de una tupla de clases), devolviendo True o False.

**Sintaxis:** isinstance(objeto, clase_o_tipo)

**Ejemplo práctico:**

```python
valor = 10.5
es_decimal = isinstance(valor, float)
print(es_decimal) # Salida: True
```

## 30. divmod()


**Descripción:** Toma dos números (dividendo y divisor) y devuelve una tupla que contiene tanto el cociente como el residuo de la división entera.

**Sintaxis:** divmod(dividendo, divisor)

**Ejemplo práctico:**

```python
cociente, residuo = divmod(10, 3)
print(f"Cociente: {cociente}, Residuo: {residuo}") # Salida: Cociente: 3, Residuo: 1
```
