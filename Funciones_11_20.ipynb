## 11. max()

**Descripción:** Devuelve el elemento de mayor valor de una colección o entre varios argumentos.

**Sintaxis:** `max(iterable)` o `max(a, b, c, ...)`

**Uso en ciencia de datos:** Encontrar el valor pico de una serie, como la venta más alta o la temperatura máxima.

```python
ventas = [1200, 3400, 980, 4500, 2100]

print("Venta más alta:", max(ventas))
print("Mayor entre 8, 15 y 3:", max(8, 15, 3))

palabras = ["datos", "python", "git"]
print("Palabra más larga:", max(palabras, key=len))
```

---
## 12. min()

**Descripción:** Devuelve el elemento de menor valor de una colección o entre varios argumentos.

**Sintaxis:** min(iterable) o min(a, b, c, ...)

**Uso en ciencia de datos:** Detectar el valor mínimo de una serie, como el precio más bajo o la nota más baja.

```python
temperaturas = [18.5, 22.0, 15.2, 25.8, 19.9]

print("Temperatura mínima:", min(temperaturas))
print("Menor entre 8, 15 y 3:", min(8, 15, 3))

vacia = []
print("Mínimo de lista vacía:", min(vacia, default="sin datos"))
```

---
## 13. abs()

**Descripción:** Devuelve el valor absoluto de un número, es decir, su distancia a cero sin importar el signo.

**Sintaxis:** abs(numero)

**Uso en ciencia de datos:** Calcular errores o diferencias entre un valor real y uno estimado sin importar si se pasó o se quedó corto.

```python
print("abs(-15):", abs(-15))
print("abs(7.3):", abs(7.3))

real = [10, 20, 30]
estimado = [12, 18, 35]

errores = [abs(r - e) for r, e in zip(real, estimado)]
print("Errores absolutos:", errores)
```

---
## 14. round()

**Descripción:** Redondea un número a la cantidad de decimales indicada. Si no se indica, redondea al entero más cercano.

**Sintaxis:** round(numero, decimales)

**Uso en ciencia de datos:** Presentar promedios y porcentajes con un número controlado de decimales.

```python
pi = 3.14159265

print("Sin decimales:", round(pi))
print("Dos decimales:", round(pi, 2))
print("Cuatro decimales:", round(pi, 4))

print("round(2.5):", round(2.5))
print("round(3.5):", round(3.5))

promedio = 17 / 3
print("Promedio redondeado:", round(promedio, 2))
```

---
## 15. sorted()

**Descripción:** Devuelve una **nueva lista** con los elementos ordenados, sin modificar la colección original.

**Sintaxis:** sorted(iterable, key=None, reverse=False)

**Uso en ciencia de datos:** Ordenar resultados para ver rankings, como los mejores puntajes o los productos más vendidos.

```python
numeros = [42, 7, 19, 3, 88]

print("Ascendente:", sorted(numeros))
print("Descendente:", sorted(numeros, reverse=True))
print("Original sin cambios:", numeros)

nombres = ["Camila", "Ana", "Sebastián", "Bea"]
print("Ordenados por longitud:", sorted(nombres, key=len))
```

---
## 16. range()

**Descripción:** Genera una secuencia de números enteros. No crea la lista completa en memoria, la produce a medida que se necesita.

**Sintaxis:** range(fin), range(inicio, fin) o range(inicio, fin, paso)

**Uso en ciencia de datos:** Repetir un proceso un número fijo de veces o generar índices.

```python
print("range(5):", list(range(5)))
print("range(2, 8):", list(range(2, 8)))
print("range(0, 20, 5):", list(range(0, 20, 5)))
print("Cuenta regresiva:", list(range(5, 0, -1)))

for i in range(1, 4):
    print("Iteración", i)
```

---
## 17. list()

**Descripción:** Crea una lista a partir de un iterable, o una lista vacía si no recibe argumentos. Las listas son ordenadas y modificables.

**Sintaxis:** list(iterable)

**Uso en ciencia de datos:** Convertir otras estructuras (texto, rangos, tuplas) en listas para poder modificarlas.

```python
vacia = list()
print("Lista vacía:", vacia)

letras = list("Python")
print("Desde un texto:", letras)

pares = list(range(0, 11, 2))
print("Desde un range:", pares)

pares.append(12)
print("Lista modificada:", pares)
```

---
## 18. tuple()

**Descripción:** Crea una tupla a partir de un iterable. Las tuplas son ordenadas pero **inmutables**: una vez creadas no se pueden modificar.

**Sintaxis:**`tuple(iterable)

**Uso en ciencia de datos:** Guardar datos que no deben cambiar, como coordenadas o registros fijos.

```python
coordenadas = tuple([4.6097, -74.0817])
print("Tupla:", coordenadas)
print("Latitud:", coordenadas[0])

try:
    coordenadas[0] = 5.0
except TypeError as error:
    print("Error esperado:", error)
```

---
## 19. dict()

**Descripción:** Crea un diccionario, una colección de pares **clave: valor**.

**Sintaxis:** dict(clave=valor, ...) o dict(lista_de_pares)

**Uso en ciencia de datos:** Representar un registro con campos con nombre, como los datos de un estudiante o un producto.

```python
estudiante = dict(nombre="Jefferson", materia="Ciencia de datos", nota=4.5)
print("Diccionario:", estudiante)
print("Nombre:", estudiante["nombre"])

pares = [("a", 1), ("b", 2), ("c", 3)]
print("Desde lista de pares:", dict(pares))

estudiante["nota"] = 4.8
print("Nota actualizada:", estudiante["nota"])
```

---
## 20. enumerate()

**Descripción:** Recorre un iterable devolviendo cada elemento junto con su posición (índice).

**Sintaxis:** enumerate(iterable, start=0)

**Uso en ciencia de datos:** Numerar registros o ubicar la posición de un dato sin manejar un contador a mano.

```python
columnas = ["fecha", "producto", "precio"]

for indice, nombre in enumerate(columnas):
    print(indice, "->", nombre)

print()
for numero, nombre in enumerate(columnas, start=1):
    print("Columna", numero, ":", nombre)
```

# Conclusiones

Las funciones max(), min() y abs() permiten comparar y medir valores; round() controla la presentación de los resultados; sorted() y enumerate() ayudan a ordenar y recorrer datos; y `range()`, `list()`, `tuple()` y `dict()` sirven para generar y organizar colecciones.

Estas funciones son la base para escribir código más claro.
