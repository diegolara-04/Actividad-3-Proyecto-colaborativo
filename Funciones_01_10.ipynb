#Funciones

## 1. print()

**Descripción:** Permite mostrar información,
mensajes y resultados en pantalla.

**Sintaxis:** print(valor)

**Ejemplo práctico:** Se utiliza para mostrar resultados de operaciones y datos procesado

```python
nombre = "Juliana"
edad = 23

print("Nombre:", nombre)
print("Edad:", edad)
```

## 2. input()

**Descripción:** Permite recibir información ingresada por el usuario mediante el teclado.

**Sintaxis:** input(mensaje)

**Ejemplo:** Se utiliza para recopilar información de los usuarios en un programa.

```python
nombre = input("Escribe tu nombre: ")

print("Hola,", nombre)
```

# 3. type()

**Descripción:** Permite identificar el tipo de dato que contiene una variable.

**Sintaxis:** type(objeto)

**Ejemplo:** Se utiliza para comprobar si un dato es entero, decimal, texto u otro tipo de información

```python
numero = 25
precio = 15.5
nombre = "Python"

print(type(numero))
print(type(precio))
print(type(nombre))
```

## 4. len()

**Descripción:** Permite conocer la cantidad de elementos de una lista o los caracteres que contiene un texto.

**Sintaxis:** len(objeto)

**Ejemplo:** Se utiliza para contar registros, elementos o caracteres.

```python
frutas = ["manzana", "pera", "uva", "mango"]

cantidad = len(frutas)

print("Cantidad de frutas:", cantidad)
```

## 5. int()

**Descripción:** Convierte un valor compatible en un número entero.

**Sintaxis:** int(valor)

**Ejemplo:** Se utiliza para convertir datos numéricos recibidos como texto en números enteros.

```python
numero = "30"

resultado = int(numero)

print("Número convertido:", resultado)
print("Tipo de dato:", type(resultado))
```

## 6. float()

**Descripción:** Convierte un valor compatible en un número decimal.

**Sintaxis:** float(valor)

**Ejemplo:** Se utiliza para trabajar con precios, temperaturas, medidas y promedios.

```python
precio = "25.75"

resultado = float(precio)

print("Precio convertido:", resultado)
print("Tipo de dato:", type(resultado))
```

## 7. str()

**Descripción:** Convierte un valor en una cadena de texto.

**Sintaxis:** str(valor)

**Ejemplo:** Se utiliza para combinar números con mensajes de texto y presentar información.

```python
edad = 22

mensaje = "Mi edad es " + str(edad)

print(mensaje)
print(type(mensaje))
```

## 8. bool()

**Descripción:** Convierte un valor en un resultado booleano, es decir, verdadero (True) o falso (False).

**Sintaxis:** bool(valor)

**Ejemplo:** Se utiliza para evaluar condiciones y comprobar si existe información.

```python
print("Número 10:", bool(10))
print("Número 0:", bool(0))
print("Texto Python:", bool("Python"))
print("Texto vacío:", bool(""))
```

## 9. sum()

**Descripción:** Permite sumar los elementos numéricos de una colección.

**Sintaxis:** sum(iterable)

**Ejemplo:** Se utiliza para calcular totales y promedios en conjuntos de datos.

```python
notas = [4.0, 3.5, 4.5, 5.0]

total = sum(notas)

promedio = total / len(notas)

print("Suma total:", total)
print("Promedio:", promedio)
```

## 10. Función set()

**Descripción:** Permite crear conjuntos de elementos únicos, eliminando los valores duplicados.

**Sintaxis:** set(iterable)

**Ejemplo:** Se utiliza para identificar valores únicos en listas y conjuntos de datos.

```python
colores = ["rojo", "azul", "rojo", "verde", "azul"]

colores_unicos = set(colores)

print("Colores sin repetir:", sorted(colores_unicos))
print("Cantidad de colores únicos:", len(colores_unicos))
```
